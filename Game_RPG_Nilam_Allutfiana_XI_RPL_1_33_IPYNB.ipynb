```python
# TUGAS 1
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
      self.name = name # Nama Hero
      self.hp = hp # Nyawa (Health Point)
      self.attack_power = attack_power # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
       print(f"Hero: {self.name} | HP: {self.hp} | Power:{self.attack_power}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Mengubah HP hero1
hero1.hp = 500
print(hero1.hp)

# Memanggil Method
hero1.info()
hero2.info()

#TUGAS 2
class Hero:
    # Constructor
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# Membuat object
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Pertarungan
print("\n--- Pertarungan Dimulai ---")

hero1.serang(hero2)  # Layla menyerang Zilong
hero2.serang(hero1)  # Zilong menyerang Layla

# TUGAS 3
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} (Mage) | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# --- Main Program Baru ---
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)           # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)   # Skill khusus Mage
```

**TUGAS ANALISIS 3**

Jika fungsi super().init(...) dihapus, maka akan muncul AttributeError: 'Mage' object has no attribute 'name'. Alasannya adalah, Child class (Mage) menimpa (override) metode inisialisasi dari Parent class (Hero). Tanpa pemanggilan super(), proses pembentukan atribut dasar seperti name, hp, dan attack_power yang berada di dalam init milik Parent class tidak akan pernah dijalankan untuk objek Mage tersebut.

Peran utamanya adalah untuk memanggil method milik Parent class dari dalam Child class. Dalam kasus ini, super() menghubungkan data inisialisasi dari Anak ke Induk untuk mencegah penulisan ulang kode pembuatan atribut yang sama.

```python
class HeroEncapsulated:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0 # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

# --- Uji Coba ---
hero1 = HeroEncapsulated("Layla", 100)
hero1.set_hp(-50)
print(hero1.get_hp())

# Uji Coba Analisis 1 (Akses Paksa)
print(f"Mencoba akses paksa: {hero1._HeroEncapsulated__hp}")
```

**TUGAS 4 ANALISIS**

Saat melakukan print(hero1._HeroEncapsulated__hp), nilai HP tetap muncul dan tidak Error. Python menggunakan mekanisme bernama Name Mangling (menambahkan _ClassName di depan atribut private). Ini dilakukan Python bukan untuk mengunci data secara absolut (seperti di Java), melainkan sekadar mencegah programmer mengubah variabel secara tidak sengaja (convention). Praktik "memaksa akses" ini tidak boleh dilakukan dalam standar pemrograman yang baik karena hal ini melanggar kontrak internal dari desain class yang sudah dibuat, yang berpotensi merusak keamanan dan stabilitas program.

Jika logika pembatasan pada setter dihapus, maka HP akan tereksekusi menjadi -100. Keberadaan method Setter sangat penting untuk menjaga "Integritas Data". Melalui Setter, programmer dapat memastikan bahwa nilai yang dimasukkan ke dalam atribut adalah masuk akal dan valid sesuai logika bisnis permainan (misal: HP tidak mungkin bernilai di bawah nol).

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Jika method ini di-comment, program akan error.
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

h = Hero("Alucard")
h.info()

# Uji Coba Analisis 2
# unit = GameUnit() # Hapus tanda pagar untuk melihat errornya.
```

**TUGAS 5 ANALSIS**

1. Melanggar Kontrak: Jika method serang di class Hero dihapus atau dikomentari, Python akan memunculkan error berbunyi: TypeError: Can't instantiate abstract class Hero with abstract method serang. Artinya, program menolak untuk mencetak objek karena class pembentuknya (Hero) belum memenuhi seluruh perjanjian kontrak standar yang ada di Interface (GameUnit). Konsekuensinya, class tersebut tidak bisa digunakan/diinstansiasi menjadi objek.

2. Mencetak Cetakan (Instansiasi Abstract Class): Saat GameUnit() coba diinstansiasi, program akan Error. Class GameUnit dilarang menjadi objek karena ia hanyalah sebuah "Konsep/Blueprint Dasar" yang tidak memiliki implementasi riil (bentuk kodenya masih pass). Gunanya ada class ini adalah sebagai "standar arsitektur/kontrak", untuk memastikan bahwa siapapun programmer yang membuat unit (entah itu monster, hero, NPC), akan memiliki method serang dan info dengan penamaan yang sama (konsisten).

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama
    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

# Child Class
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# --- UJI COBA ANALISIS 1: MENAMBAHKAN HEALER ---
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")

# Penerapan Polymorphism
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Estes") # Healer masuk list tanpa mengubah core looping
]

print("--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

**TUGAS 6 ANALISIS**

**1. Uji Skalabilitas:**
Saat objek Healer ditambahkan, program berjalan sangat lancar. Keuntungan polimorfisme adalah pada fleksibilitas (Open/Closed Principle): kita bisa dengan sangat mudah menambahkan fitur, karakter, atau mekanik baru di masa depan tanpa perlu membongkar struktur perulangan (looping utama) programnya.

**2. Konsistensi Penamaan:**
Jika nama method serang milik Archer diubah menjadi tembak_panah, program akan memunculkan error AttributeError pada gilirannya si objek Archer. Ini dikarenakan struktur Looping menuntut instruksi pemanggilan yang seragam, yaitu pahlawan.serang(). Karena itulah dalam polimorfisme nama method antara Parent dan seluruh Child Class harus persis sama agar bisa dipanggil serentak di satu pintu tanpa program harus memeriksa satu per satu "tipe pahlawan ini menggunakan method apa".
